## 1. Groundwater flow

The flow of water through a groundwater aquifer and the flow into a nearby river can be modeled by combining Darcy's law for the flow thorugh a porous medium with consevation of mass. This leads to a partial differential equation for the height of the water table, $h(x,t)$, where $x$ is the distance from the river and $t$ is time.


### 1.1 Theory

The theory for the flow through a groundwater aquifer is briefly summarized here. See the FEM lectures and notes for a full explanation. Consider two-dimensional flow of groundwater toward a river. Let $x$ be the distance from the river, $h(x,t)$ be the height of the water table, $\phi$ be the porosity of the aquifer, and $R(t)$ be the rainfall rate.

Darcy's law balances viscous resistance in the porous medium against pressure gradients (including buoyancy). For a long, thin aquifer the vertical velocity is small, so the pressure is approximately hydrostatic. The horizontal velocity in the aquifer is then

$$ u = -u_b \frac{\partial h}{\partial x}, $$

where $u_b= k\rho g/\mu$ is the buoyancy velocity, $k$ is the permeability, $\rho$ is the fluid density, $g$ is gravity, and $\mu$ is the dynamic viscosity.

The corresponding volume flux (per unit width) through a vertical section of the aquifer is

$$ q = \int_0^h u\,\mathrm{d}z = -u_b h \frac{\partial h}{\partial x}, $$

assuming that the buoyancy velocity is constant. Conservation of mass then gives an evolution equation for the height of the water table:

$$ \phi \frac{\partial h}{\partial t} - u_b \frac{\partial}{\partial x}\left(h\frac{\partial h}{\partial x}\right) = R. $$

The boundary conditions are $h=0$ at the river ($x=0$) and no flux through the drainage divide ($q=0$ at $x=L$). The volume flux into the river is

$$ Q_R = -q(x=0) = u_b h \frac{\partial h}{\partial x}\bigg|_{x=0}. $$



### 1.2 Model

The script below configures and runs Dedalus to solve for the height of the groundwater aquifer and the volume flux into the river. In this example, the rainfall rate is

$$ R(t) = \begin{cases} 0 & t < 0, \\ R_0 & 0 < t < T, \\ 0 & t > T, \end{cases} $$

where the rainfall rate is steady for a time $T$.

For simplicity, we will assume that the aquifer is empty before the rain so that

$$ h(t=0) = 0. $$

Since the boundary conditions at $x=0$ and $x=L$ are not periodic, we solve the equation using a Chebyshev transform in the $x$ direction.

```python
"""
Script to use Dedalus to solve for the height of the water table and the volume flux into the river
"""

import numpy as np
import matplotlib.pyplot as plt
import dedalus.public as dedalus
import logging
logger = logging.getLogger(__name__)

# For convenience, define 1 hour and 1 day
hour = 3600 # seconds
day = 24 * hour
kilometer = 1e3

# Set parameters
phi = 0.3 # porosity (non-dimensional)
k = 1e-10 # permeability, units: m^2
rho = 1000 # density of water in kg/m^3
g = 9.81 # gravitational acceleration (m/s^2)
mu = 1e-3 # dynamic viscosity of water in kg/(m*s)
ub = k * rho * g / mu # buoyancy velocity (m/s)
R0 = 5e-3 / day # rainfall rate (5 mm/day)
T = 60 * day # rainfall duration
L = 100 # length of the aquifer (m)

# Derived parameters
H = L*(R0/ub)**0.5 # water table height at steady state

# Numerical parameters
Nx = 64 # number of gridpoints in x
stop_sim_time = 120 * day # end time
timestep = 1 * hour # timestep
nu = ub * (R0/ub)**0.5 * L # implicit diffusivity based on the characteristic water table height

# Bases and coordinates
coords = dedalus.CartesianCoordinates('x')
dist = dedalus.Distributor(coords, dtype=np.float64)
xbasis = dedalus.ChebyshevT(coords['x'], size=Nx, bounds=(0, L), dealias=3/2) # what does this do?
x = dist.local_grid(xbasis)

# Fields
h = dist.Field(name='h', bases=xbasis)
tau_1 = dist.Field(name='tau_1')
tau_2 = dist.Field(name='tau_2')
R = dist.Field(name='R')
R['g'] = R0

# Substitutions
lift_basis = xbasis.derivative_basis(1)
lift = lambda A: dedalus.Lift(A, lift_basis, -1)
dx = lambda A: dedalus.Differentiate(A, coords['x'])
hx = dx(h) + lift(tau_1) # define the derivative of h(x)

# Problem
problem = dedalus.IVP([h, tau_1, tau_2], namespace=locals())
problem.add_equation("phi*dt(h) - nu*dx(hx) + lift(tau_2) = R + ub*dx(h*hx) - nu*dx(hx)")
problem.add_equation("h(x='left') = 0")
problem.add_equation("hx(x='right') = 0")

# Now, set the solver
solver = problem.build_solver(dedalus.RK222)
solver.stop_sim_time = stop_sim_time

# Initial condition: empty aquifer
h['g'] = 0

# Volume of water in the aquifer (per unit width)
volume = phi * dedalus.integ(h)

# River flux Q_R = ub * h * dh/dx evaluated at the river
QR = (ub * h * hx)(x='left')

# Create arrays to periodically save the water table height for plotting later
h.change_scales(1)
h_save = [np.copy(h['g'])];
t_save = [solver.sim_time]; # Save the initial condition and the initial time
volume_save = [np.copy(volume.evaluate()['g'])]
QR_save = [np.copy(QR.evaluate()['g'])]

# Specify the rainfall rate as a function of time
def rainfall_rate(t):
    #return R0 * np.ones_like(t)
    return np.where((t >= 0) & (t < T), R0, 0.0)

# Main loop
try:
    logger.info('Starting main loop')
    while solver.proceed:
        R['g'] = rainfall_rate(solver.sim_time)
        solver.step(timestep)
        if (solver.iteration-1) % 10 == 0:
            #logger.info('Iteration=%i, Time=%e days, dt=%e' %(solver.iteration, solver.sim_time/day, timestep))
            h.change_scales(1)
            h_save.append(np.copy(h['g']))
            t_save.append(solver.sim_time)
            volume_save.append(np.copy(volume.evaluate()['g']))
            QR_save.append(np.copy(QR.evaluate()['g']))
except:
    logger.error('Exception raised, triggering end of main loop.')
    raise
finally:
    solver.log_stats()

# Convert the variables to numpy arrays for allow array slicing
h_save = np.array(h_save)
t_save = np.array(t_save)
volume_save = np.array(volume_save).squeeze()
QR_save = np.array(QR_save).squeeze()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert time to days and flux to m^2/day
t_days = t_save / day
flux_days = QR_save * day

# --- 1. Linear Fit for Early Times ---
# We fit a line (y = m*t) to the initial rising phase (first 30 days)
early_mask = (t_days > 0) & (t_days <= 30)
t_early = t_days[early_mask]
flux_early = flux_days[early_mask]

slope = np.sum(t_early * flux_early) / np.sum(t_early**2)
linear_fit = slope * t_days

# --- 2. Fit for Late Times (Power Law Decay 1/t^2) ---
# Fit the expected analytical scaling form: Q = alpha / t^2
# We fit this to the decaying phase after rainfall stops (t >= 60 days)
late_mask = t_days >= 60
t_late = t_days[late_mask]
flux_late = flux_days[late_mask]

# Solve for alpha in the least-squares sense: flux_late = alpha * (1 / t_late^2)
x_late = 1.0 / (t_late**2)
alpha = np.sum(x_late * flux_late) / np.sum(x_late**2)
power_law_fit = alpha / (t_days**2)

# --- Plotting ---
plt.figure(figsize=(10, 6))
plt.plot(t_days, flux_days, 'k-', label='Simulation Flux', linewidth=2)
plt.plot(t_days[t_days <= 60], linear_fit[t_days <= 60], 'r--', label=f'Early Linear Fit ($Q \propto t$, Slope: {slope:.2e})', linewidth=1.5)
plt.plot(t_days[t_days >= 50], power_law_fit[t_days >= 50], 'b--', label=f'Late Analytical Fit ($Q \propto t^{{-2}}$, alpha: {alpha:.2e})', linewidth=1.5)

plt.xlim(0, stop_sim_time / day)
plt.ylim(0, np.max(flux_days) * 1.1)
plt.xlabel('Time (days)')
plt.ylabel('Flux into river (m$^2$/day)')
plt.title('Groundwater Flux into River with Early (Linear) and Late ($1/t^2$) Fits')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

For early times, the flux into the river $Q ∼ \frac{\sqrt{u_bR^3}}{ϕ}t$, giving a linear relationship between flux and time. At very early times ($t < 10$), the simulation increases at a slower rate than the linear prediction as the numerical solver takes some time to get used to the steep gradient at $t=0$. Reducing the timestep reduces this difference. The simulation is then approximately linear, but as time becomes larger the simulation output begins to diverge from the prediction as it tends towards a steady state. Our scaling analysis for the initial flux makes the assumption that the height of the aquifer $h$ scales with $\frac{Rt}{ϕ}$ as we consider the aquifer flat in the far field and therefore constant w.r.t $x$. As the aquifer grows, the divergence of the fluid flux term begins to have a greater effect and therefore $h$ no longer scales linearly with time $t$.

At later times, the flux scales like $Q ∼ \frac{ϕ^2 L ^3}{u_b}\frac{1}{t^2}$; the flux is inversely proportional to time squared. This is because the height of the aquifer now varies significantly in $x$, so it scales like $h\sim \frac{L^2 ϕ}{u_b t}$. This scaling is true as $t → ∞$, but initially after the rain stops there are other contributing terms that mean the aquifer height does not purely decay with $t^{-2}$.

Explore different functional forms for the rainfall rate and the properties of the reservoir (e.g. size and permeability).  Include 2-3 figures to show the resutls of your parameter exploration and add 1-2 pargraphs to discuss and interpret your results.

```python
def run_aquifer_simulation(k_val=1e-9, R0_val=5e-3/day, rain_type='constant', L_val=100.0, stop_time=120*day, rain_duration=60*day):
    """Runs Dedalus for given parameters and returns time series results"""
    import dedalus.public as dedalus
    import numpy as np

    # Physical parameters
    phi_val = 0.3
    rho_val = 1000.0
    g_val = 9.81
    mu_val = 1e-3
    ub_val = k_val * rho_val * g_val / mu_val

    # Numerical parameters
    Nx_val = 64
    if k_val >= 1e-6:
        timestep_val = 0.5 * hour
    else:
        timestep_val = 1 * hour

    nu_val = ub_val * (abs(R0_val)/max(ub_val, 1e-12))**0.5 * L_val if ub_val > 0 else 1e-3
    if nu_val == 0:
        nu_val = 1e-3

    coords = dedalus.CartesianCoordinates('x')
    dist = dedalus.Distributor(coords, dtype=np.float64)
    xbasis = dedalus.ChebyshevT(coords['x'], size=Nx_val, bounds=(0, L_val), dealias=3/2)
    x_grid = dist.local_grid(xbasis)

    h_f = dist.Field(name='h', bases=xbasis)
    tau_1_f = dist.Field(name='tau_1')
    tau_2_f = dist.Field(name='tau_2')
    R_f = dist.Field(name='R')
    R_f['g'] = R0_val

    lift_basis = xbasis.derivative_basis(1)
    lift_op = lambda A: dedalus.Lift(A, lift_basis, -1)
    dx_op = lambda A: dedalus.Differentiate(A, coords['x'])
    hx_f = dx_op(h_f) + lift_op(tau_1_f)

    problem = dedalus.IVP([h_f, tau_1_f, tau_2_f], namespace=locals())
    problem.add_equation("phi_val*dt(h_f) - nu_val*dx_op(hx_f) + lift_op(tau_2_f) = R_f + ub_val*dx_op(h_f*hx_f) - nu_val*dx_op(hx_f)")
    problem.add_equation("h_f(x='left') = 0")
    problem.add_equation("hx_f(x='right') = 0")

    solver = problem.build_solver(dedalus.RK222)
    solver.stop_sim_time = stop_time
    h_f['g'] = 0.0

    # For noise generation
    np.random.seed(42)

    def get_R(t):
        if t < 0 or t > rain_duration:
            return 0.0
        if rain_type == 'constant':
            return R0_val
        elif rain_type == 'lower':
            return R0_val * 0.5
        elif rain_type == 'noisy':
            # Uniform distribution U(0, 10) yields an expected average of 5 mm/day
            noise_mm_day = np.random.uniform(0.0, 10.0)
            return (noise_mm_day / 1e3) / day
        return R0_val

    t_out = []
    R_out = []
    h_center_out = []
    QR_out = []

    QR_expr = (ub_val * h_f * hx_f)(x='left')

    while solver.proceed:
        current_R = get_R(solver.sim_time)
        R_f['g'] = current_R
        solver.step(timestep_val)

        if solver.iteration % 10 == 0 or solver.sim_time >= solver.stop_sim_time:
            h_f.change_scales(1)
            t_out.append(solver.sim_time)
            R_out.append(current_R)
            h_center_out.append(h_f['g'][Nx_val // 2])
            QR_out.append(QR_expr.evaluate()['g'][0])

    return np.array(t_out), np.array(R_out), np.array(h_center_out), np.array(QR_out)
```

```python
# 1. Permeability Investigation (k = 1e-12, 1e-9, 1e-6 m^2) - Rain stops at 60 days
ks = [1e-12, 1e-9, 1e-6]
k_names = ['Clay ($10^{-12}$ $m^2$)', 'Silt ($10^{-9}$ $m^2$)', 'Sand ($10^{-6}$ $m^2$)']
results_k = []

for k_val in ks:
    t_arr, R_arr, hc_arr, QR_arr = run_aquifer_simulation(k_val=k_val, R0_val=5e-3/day, rain_type='constant', L_val=100.0, stop_time=120*day, rain_duration=60*day)
    results_k.append((t_arr, R_arr, hc_arr, QR_arr))

# Create 3x3 plot for Permeability Investigation
fig, axes = plt.subplots(3, 3, figsize=(14, 12), sharex='col')
for col, (name, res) in enumerate(zip(k_names, results_k)):
    t_days = res[0] / day
    R_mm = res[1] * 1e3 * day
    hc = res[2]
    QR_flux = res[3] * day

    # Row 1: Rainfall Rate
    axes[0, col].plot(t_days, R_mm, color='tab:blue')
    axes[0, col].set_title(name, fontsize=12)
    axes[0, col].set_ylabel('Rainfall (mm/day)')
    axes[0, col].grid(True, linestyle=':', alpha=0.6)

    # Row 2: Center Aquifer Height
    axes[1, col].plot(t_days, hc, color='tab:green')
    axes[1, col].set_ylabel('Center h (m)')
    axes[1, col].grid(True, linestyle=':', alpha=0.6)

    # Row 3: Flux into River
    axes[2, col].plot(t_days, QR_flux, color='tab:red')
    axes[2, col].set_ylabel('River Flux ($m^2$/day)')
    axes[2, col].set_xlabel('Time (days)')
    axes[2, col].grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Changing Permeability', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

Low permeability ground, such as clay, holds a very high amount of water in the aquifer, and drains very slowly, giving a small river flux. High permeability ground, such as sand, allows the water to drain very quickly, meaning the river flux responds to changes in rainfall over very short timescales, and the aquifer height barely increases as any 'rainfall in' almost immediately drains out.

## 2. Transport of pollution in a river

Advection/diffusion equations are a common way to model the material transport in many physical systems. In this section we will model the transport and dilution of a pollutants added to a river.

For simplicity, we will model the river as a rectangular channel, and we will neglect depth variations. Let $x$ be the coordinate along the length of the river, and $y$ be the cross-stream coordinate. We will model the river current as

$$ u(y) = U_0\left(1-\frac{(y-w/2)^2}{(w/2)^2}\right) $$

where $w$ is the width of the river and $U_0$ is the maximum current speed which occurs at the center of the river (the $y$-coordinate will run from 0 to $w$).

Since we want to know how the pollution levels vary in time, we will solve the time-dependent advection/diffusion equation. We will model the addition of pollution by adding a spatially-dependent source term to the right hand side of the equation governing pollutant concentration:

$$ \frac{\partial c}{\partial t} + u(y)\frac{\partial c}{\partial x} = \kappa \nabla^2 c + S(x,y) $$

where $\nabla^2=(\partial_x^2 + \partial_y^2)$ is the Laplacian operator for diffusion in 2D Cartesian coordinates, and $S$ is the source term.

### 2.1 Dedalus script

The script below solves the time-dependent advection/diffusion equation with a source term.

```python
"""
Script to solve the time-dependent advection/diffusion equation
"""

import numpy as np
import matplotlib.pyplot as plt
import dedalus.public as dedalus
import logging
logger = logging.getLogger(__name__)

# Set parameters
U0 = 0.01 # maximum current speed in m/s
w = 10 # width of the river in m
L = 50 # length of the river section to simulate
kappa = 1e-2 # diffusivity in m^2/s
S0 = 1 # amplitude of the pollution source term
S_w = 1 # width of the source term in m
S_x = 10 # center of the forcing in x
S_y = 0 # center of the forcing in y
S_t = 1 # forcing timescale

# Numerical parameters
Nx = 256 # number of gridoints in x
Ny = 32 # number of gridpoints in y
stop_sim_time=1500 # end time in seconds
timestep = (L/Nx)/U0 # timestep in seconds, calculated from the CFL number

import numpy as np
import dedalus.public as dedalus
import logging
logger = logging.getLogger(__name__)

# Bases and coordinates
coords = dedalus.CartesianCoordinates('x', 'y')
dist = dedalus.Distributor(coords, dtype=np.float64)
xbasis = dedalus.RealFourier(coords['x'], size=Nx, bounds=(0, L), dealias=3/2)
ybasis = dedalus.ChebyshevT(coords['y'], size=Ny, bounds=(0, w), dealias=3/2)
x, y = dist.local_grids(xbasis, ybasis)
ex, ey = coords.unit_vector_fields(dist) # x and y unit vectors

# Fields
c = dist.Field(name='c', bases=(xbasis,ybasis))
u = dist.Field(name='u', bases=(xbasis,ybasis))
tau_1 = dist.Field(name='tau_1', bases=xbasis)
tau_2 = dist.Field(name='tau_2', bases=xbasis)

# Specify the velocity field
u['g'] = U0 * (1- (y - w/2)**2.0/(w/2)**2.0) # parabolic velocity profile

# Substitutions
lift_basis = ybasis.derivative_basis(1)
lift = lambda A: dedalus.Lift(A, lift_basis, -1)
grad_c = dedalus.grad(c) + ey*lift(tau_1)
dx = lambda A: dedalus.Differentiate(A, coords['x'])

S = dist.Field(name='u', bases=(xbasis,ybasis))
S['g'] = S0*np.exp(-(x-S_x)**2.0/S_w**2.0-(y-S_y)**2.0/S_w**2.0)

# Problem
problem = dedalus.IVP([c, tau_1, tau_2], namespace=locals())
problem.add_equation("dt(c) + u*dx(c) - kappa*div(grad_c) + lift(tau_2) = S/S_t")
problem.add_equation("c(y=0) = 0")
problem.add_equation("c(y=w) = 0")

# Now, set the solver
solver = problem.build_solver(dedalus.RK222)
solver.stop_sim_time = stop_sim_time

# Create an array to periodically save the concentration field for plotting later
c.change_scales(1)
c_save = [np.copy(c['g'])];

t_save = [solver.sim_time]; # Save the initial condition and the initial time
# Main loop
try:
    logger.info('Starting main loop')
    while solver.proceed:
        solver.step(timestep)
        if (solver.iteration-1) % 10 == 0:
            #logger.info('Iteration=%i, Time=%e, dt=%e' %(solver.iteration, solver.sim_time, timestep))
            c.change_scales(1)
            c_save.append(np.copy(c['g']))
            t_save.append(solver.sim_time)
except:
    logger.error('Exception raised, triggering end of main loop.')
    raise
finally:
    solver.log_stats()

# Convert the variables to numpy arrays for allow array slicing
c_save = np.array(c_save)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from dedalus.extras import plot_tools

# Target times to plot
target_times = [10, 200, 500, 1500]

# Create 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
axes = axes.flatten()

# Construct coordinates mesh
(X_mesh, Y_mesh) = plot_tools.quad_mesh(np.squeeze(x), np.squeeze(y))

for i, t_val in enumerate(target_times):
    # Find the index closest to the target time in t_save
    idx = np.argmin(np.abs(np.array(t_save) - t_val))
    actual_time = t_save[idx]

    # Plot using pcolormesh
    im = axes[i].pcolormesh(X_mesh, Y_mesh, c_save[idx, :, :].T, shading='auto')

    # Formatting
    axes[i].set_title(f'Time t = {actual_time:.1f} s')
    axes[i].set_xlim(0, L)
    axes[i].set_ylim(0, w)
    axes[i].set_aspect('equal')
    if i >= 2:
        axes[i].set_xlabel('x (m)')
    if i % 2 == 0:
        axes[i].set_ylabel('y (m)')

# Add a colorbar to show concentration scale
fig.subplots_adjust(right=0.85, hspace=0.3, wspace=0.1)
cbar_ax = fig.add_axes([0.88, 0.15, 0.03, 0.7])
fig.colorbar(im, cax=cbar_ax, label='Concentration c')

plt.suptitle('Pollution Concentration Evolution over Time', fontsize=14)
plt.show()
```

The timescale $\tau$ of the pollutant reaching the other side of the river scales like $\tau \sim \frac{w^2}{\kappa}$, where $w$ is the width of the river and $\kappa$ is the diffusivity. Using this timescale, the length $L_x$ downstream where the pollutant will reach the other side scales like $L_x \sim U_0 \tau \sim U_0 \frac{w^2}{\kappa}$. This matches intuition and observations in the model - a larger diffusivity will mean the pollutant reaches the other side faster, and will therefore shorten the distance travelled by the pollutant downstream. A larger current $U_0$ in the stream will not affect the time it takes for the pollutant to reach the other side, however it will affect how far the pollutant travels in the stream before reaching the other side.

In the above plot, $U_0 = 0.01$ and $\kappa = 0.1$, which gives $\tau ≈ 1000$ s and $L_x ≈ 10$ m. This doesn't appear to be strictly accurate; the timescale is approximately one order of magnitude out, and the lengthscale seems smaller than expected. An explanation for this could be that $U_0$ is used in the scaling, which is the maximum velocity, however the actual velocity in the river follows a parabolic profile and therefore the average downstream velocity as the pollutant diffuses across the river is less than $U_0$. I'm not sure where the timescale error comes from.

### Turbulent diffusion
In a turbulent flow, tracers will be transported by the turbulent eddies. A common way to model this is to introduce a turbulent diffusivity, $\kappa_T\sim u*l$, where $u$ and $l$ are characteristic velocity and length scales associated with the turbulent eddies.

To explore this effect, create a Dedalus script that includes advection by $u$ and $v$ (the $x$ and $y$ components of the velocity). Then, use the same velocity field as above for $u$ and add to that a fluctuating velocity field that of the form

$$ u=u_0 cos(kx)cos(ky), \quad v=u_0 sin(kx)sin(ky) $$

where $k$ is the wavenumber associated with the velocity fluctuations and $u_0$ is the amplitude of the fluctuations. Note that $k$ should be set to give an integer number of wavelengths in the $x$ and $y$ directions. Vary both $u_0$ and $k$ and study how quicky the tracer is transported across the river. Qualitatively compare this with the expectation based on the hypothesis of turbulent diffusion.

```python
import numpy as np
import matplotlib.pyplot as plt
import dedalus.public as dedalus
import logging
from dedalus.extras import plot_tools
logging.disable(logging.CRITICAL)

# Set parameters
U0 = 0.01 # maximum current speed in m/s
w = 10 # width of the river in m
L = 50 # length of the river section
kappa = 0.01 # background molecular diffusivity in m^2/s
S0 = 1 # amplitude of the source term
S_w = 1 # width of the source
S_x = 10 # source x-center
S_y = 1 # source y-center
S_t = 1 # forcing timescale

# Turbulent fluctuation parameters
u0 = 0.2 # amplitude of velocity fluctuations (try varying this)
# Wavenumber k to give an integer number of wavelengths:
# e.g., 2 wavelengths in y over width w=10 implies wavelength lambda_y = 5m, so k_y = 2*pi/5
# To maintain isotropic eddies, we choose k = 2 * np.pi / 5.0
k_fluc = 1.0 * np.pi / 5

# Numerical parameters
Nx = 256
Ny = 32
stop_sim_time = 5000
timestep = (L/Nx)/max(U0, u0)

# Bases and coordinates
coords = dedalus.CartesianCoordinates('x', 'y')
dist = dedalus.Distributor(coords, dtype=np.float64)
xbasis = dedalus.RealFourier(coords['x'], size=Nx, bounds=(0, L), dealias=3/2)
ybasis = dedalus.ChebyshevT(coords['y'], size=Ny, bounds=(0, w), dealias=3/2)
x_g, y_g = dist.local_grids(xbasis, ybasis)
ey = coords.unit_vector_fields(dist)[1]

# Fields
c = dist.Field(name='c', bases=(xbasis, ybasis))
u = dist.Field(name='u', bases=(xbasis, ybasis))
v = dist.Field(name='v', bases=(xbasis, ybasis))
tau_1 = dist.Field(name='tau_1', bases=xbasis)
tau_2 = dist.Field(name='tau_2', bases=xbasis)

# Prescribe parabolic background velocity + fluctuation field
u['g'] = U0 * (1.0 - (y_g - w/2.0)**2.0 / (w/2.0)**2.0) + u0 * np.cos(k_fluc * x_g) * np.cos(k_fluc * y_g)
v['g'] = u0 * np.sin(k_fluc * x_g) * np.sin(k_fluc * y_g)

# Source term
S = dist.Field(name='S', bases=(xbasis, ybasis))
S['g'] = S0 * np.exp(-(x_g - S_x)**2.0 / S_w**2.0 - (y_g - S_y)**2.0 / S_w**2.0)

# Substitutions
lift_basis = ybasis.derivative_basis(1)
lift = lambda A: dedalus.Lift(A, lift_basis, -1)
grad_c = dedalus.grad(c) + ey*lift(tau_1)
dx = lambda A: dedalus.Differentiate(A, coords['x'])
dy = lambda A: dedalus.Differentiate(A, coords['y'])

# IVP formulation
problem = dedalus.IVP([c, tau_1, tau_2], namespace=locals())
problem.add_equation("dt(c) + u*dx(c) + v*dy(c) - kappa*div(grad_c) + lift(tau_2) = S/S_t")
problem.add_equation("c(y=0) = 0")
problem.add_equation("c(y=w) = 0")

solver = problem.build_solver(dedalus.RK222)
solver.stop_sim_time = stop_sim_time

c.change_scales(1)
c_save_turb = [np.copy(c['g'])]
t_save_turb = [solver.sim_time]

# Run simulation
while solver.proceed:
    solver.step(timestep)
    if (solver.iteration - 1) % 10 == 0:
        c.change_scales(1)
        c_save_turb.append(np.copy(c['g']))
        t_save_turb.append(solver.sim_time)

c_save_turb = np.array(c_save_turb)
t_save_turb = np.array(t_save_turb)
```

```python
# Plot the concentration profiles for Turbulent Diffusion at 4 target times
target_times = [10, 200, 500, 1500]
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
axes = axes.flatten()

(X_mesh, Y_mesh) = plot_tools.quad_mesh(np.squeeze(x_g), np.squeeze(y_g))

for i, t_val in enumerate(target_times):
    idx = np.argmin(np.abs(t_save_turb - t_val))
    actual_time = t_save_turb[idx]

    im = axes[i].pcolormesh(X_mesh, Y_mesh, c_save_turb[idx, :, :].T, shading='auto', cmap='viridis')
    axes[i].set_title(f'Turbulent flow, t = {actual_time:.1f} s')
    axes[i].set_xlim(0, L)
    axes[i].set_ylim(0, w)
    axes[i].set_aspect('equal')
    if i >= 2:
        axes[i].set_xlabel('x (m)')
    if i % 2 == 0:
        axes[i].set_ylabel('y (m)')

fig.subplots_adjust(right=0.85, hspace=0.3, wspace=0.1)
cbar_ax = fig.add_axes([0.88, 0.15, 0.03, 0.7])
fig.colorbar(im, cax=cbar_ax, label='Concentration c')

plt.suptitle('Turbulent Diffusion: Tracer Concentration Evolution ($u_0 = 0.2$, $k = 0.63$)', fontsize=14, fontweight='bold')
plt.show()
```

The four images below show combinations of high/low $u_0$ and $k$, with each respective subplot at a similar timestep. It is clear that a larger amplitude of velocity fluctuations $u_0$ results in the tracer being transported more quickly, which makes sense as if the rotation of the eddies is faster, the tracer will be transported between eddies faster. The wavenumber $k$ has an inversely proportional relationship with the speed at which the tracer moves across the river; smaller wavenumber leads to larger eddies, which transport the tracer more quickly across the river.

## 3. Optional further investigations

* In the river pollution example, you might notice that the concentration field is mixed more effectively in the x-direction than in the y-direction. The combination of differential advection by the currents and cross-stream diffusion is very effective at mixing the plume in the x-direction. This is an example of Taylor dispersion (Taylor, 1953). Construct numerical experiments to quantify Taylor dispersion and its dependence on the current shear and the diffusivity. You could also separate the diffusivity into parts in the x and y directions with different coefficients.  If you set the x-component of the diffusivity to zero, then any spreading of the concentration in the x-direction should be due to Taylor dispersion.

* Try making the river speed a function of time. The easiest way to do this is to add a new equation so that Dedalus solves for the velocity in addition to the concentration field. For example, you could make the river speed decrease exponentially in time, or it could be sinusoidal in time.
